# Algorand Holistic Loan Approval Decision Engine

This notebook demonstrates the comprehensive governance reputation and network risk monitoring engines for holistic loan approval decisions.

## Features Demonstrated:
- **Governance Reputation Analysis**: Long-term ecosystem commitment evaluation
- **Network Risk Monitoring**: Real-time ecosystem risk assessment
- **Holistic Decision Making**: Integrated approach considering borrower's ecosystem role
- **Dynamic Risk Adjustment**: Real-time parameter adjustments based on network conditions

```python
# Import required libraries
import asyncio
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# Add parent directory to path
sys.path.insert(0, str(Path('.').parent))

# Configure plotting
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
%matplotlib inline
```

## 1. Initialize Engines

```python
from governance_reputation.core.reputation_engine import ReputationEngine
from network_risk_monitoring.core.risk_engine import NetworkRiskEngine
from decision_engine import LoanDecisionEngine, LoanApplication

# Initialize engines
reputation_engine = ReputationEngine()
risk_engine = NetworkRiskEngine()
base_engine = LoanDecisionEngine()

print("✅ Engines initialized successfully!")
```

## 2. Governance Reputation Analysis

```python
# Analyze governance reputation for a sample borrower
sample_address = "ALGORAND_ECOSYSTEM_CONTRIBUTOR_12345"

async def analyze_governance_reputation():
    print("🏛️ Analyzing Governance Reputation...")
    
    reputation_analysis = await reputation_engine.analyze_comprehensive_reputation(sample_address)
    
    # Display key metrics
    print(f"\n📊 Overall Reputation Score: {reputation_analysis.overall_reputation_score:.1f}")
    print(f"🏆 Reputation Tier: {reputation_analysis.reputation_tier.value.upper()}")
    
    # Component scores
    component_scores = {
        'Governance': reputation_analysis.governance_score,
        'Community': reputation_analysis.community_engagement_score,
        'Consensus': reputation_analysis.consensus_participation_score,
        'Commitment': reputation_analysis.ecosystem_commitment_score
    }
    
    return reputation_analysis, component_scores

reputation_result, rep_scores = await analyze_governance_reputation()
```

```python
# Visualize reputation component scores
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Component scores radar chart
components = list(rep_scores.keys())
scores = list(rep_scores.values())

ax1.bar(components, scores, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
ax1.set_title('Governance Reputation Component Scores', fontsize=14, fontweight='bold')
ax1.set_ylabel('Score (0-100)')
ax1.set_ylim(0, 100)

# Add score labels on bars
for i, score in enumerate(scores):
    ax1.text(i, score + 2, f'{score:.1f}', ha='center', fontweight='bold')

# Loan decision impact
impact = reputation_result.loan_decision_impact
impact_data = {
    'LTV Adjustment': impact.ltv_adjustment * 100,
    'Rate Adjustment': impact.rate_adjustment * 100,
    'Approval Bias': impact.approval_bias * 100
}

colors = ['green' if v > 0 else 'red' if v < 0 else 'gray' for v in impact_data.values()]
ax2.bar(impact_data.keys(), impact_data.values(), color=colors)
ax2.set_title('Reputation Impact on Loan Decisions', fontsize=14, fontweight='bold')
ax2.set_ylabel('Adjustment (%)')
ax2.axhline(y=0, color='black', linestyle='-', alpha=0.3)

plt.tight_layout()
plt.show()

print("📈 Reputation analysis visualization complete!")
```

## 3. Network Risk Monitoring

```python
# Perform comprehensive network risk assessment
async def analyze_network_risk():
    print("🌐 Analyzing Network Risk...")
    
    risk_assessment = await risk_engine.assess_comprehensive_risk()
    
    print(f"\n📊 Overall Risk Score: {risk_assessment.overall_risk_score:.1f}")
    print(f"🚨 Risk Level: {risk_assessment.risk_level.value.upper()}")
    
    return risk_assessment

risk_result = await analyze_network_risk()
```

```python
# Visualize network risk components
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Risk component scores
risk_components = list(risk_result.component_scores.keys())
risk_scores = list(risk_result.component_scores.values())

# Create color map based on risk level
colors = ['red' if score > 70 else 'orange' if score > 40 else 'green' for score in risk_scores]

ax1.barh(risk_components, risk_scores, color=colors)
ax1.set_title('Network Risk Component Scores', fontsize=14, fontweight='bold')
ax1.set_xlabel('Risk Score (0-100)')
ax1.set_xlim(0, 100)

# Add score labels
for i, score in enumerate(risk_scores):
    ax1.text(score + 1, i, f'{score:.1f}', va='center', fontweight='bold')

# Risk adjustments pie chart
adjustments = risk_result.risk_adjustments
adjustment_labels = ['LTV Reduction', 'Rate Increase', 'Threshold Increase']
adjustment_values = [
    abs(adjustments.ltv_adjustment) * 100,
    abs(adjustments.interest_rate_adjustment) * 100,
    abs(adjustments.approval_threshold_adjustment) * 100
]

# Only show non-zero adjustments
non_zero = [(label, value) for label, value in zip(adjustment_labels, adjustment_values) if value > 0]
if non_zero:
    labels, values = zip(*non_zero)
    ax2.pie(values, labels=labels, autopct='%1.1f%%', startangle=90)
    ax2.set_title('Risk-Based Loan Adjustments', fontsize=14, fontweight='bold')
else:
    ax2.text(0.5, 0.5, 'No Risk Adjustments\nRequired', ha='center', va='center', 
             transform=ax2.transAxes, fontsize=14, fontweight='bold')
    ax2.set_title('Risk-Based Loan Adjustments', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

print("📈 Risk analysis visualization complete!")
```

## 4. Holistic Loan Decision Demo

```python
# Create sample loan applications
loan_applications = [
    {
        'name': 'Alice - Ecosystem Contributor',
        'app': LoanApplication(
            loan_id="LOAN_001",
            borrower_name="Alice Ecosystem Contributor",
            borrower_address="ALICE_ALGO_ADDRESS",
            amount=50000.0,
            duration_days=365,
            purpose="DeFi protocol development",
            credit_score=750,
            annual_income=120000.0,
            debt_to_income=0.25,
            collateral_value=75000.0,
            employment_years=5,
            previous_defaults=0
        )
    },
    {
        'name': 'Bob - Average User',
        'app': LoanApplication(
            loan_id="LOAN_002",
            borrower_name="Bob Average User",
            borrower_address="BOB_ALGO_ADDRESS",
            amount=25000.0,
            duration_days=180,
            purpose="Personal investment",
            credit_score=680,
            annual_income=75000.0,
            debt_to_income=0.35,
            collateral_value=30000.0,
            employment_years=3,
            previous_defaults=0
        )
    },
    {
        'name': 'Charlie - High Risk',
        'app': LoanApplication(
            loan_id="LOAN_003",
            borrower_name="Charlie High Risk",
            borrower_address="CHARLIE_ALGO_ADDRESS",
            amount=100000.0,
            duration_days=730,
            purpose="Speculative trading",
            credit_score=580,
            annual_income=60000.0,
            debt_to_income=0.45,
            collateral_value=80000.0,
            employment_years=1,
            previous_defaults=2
        )
    }
]

print(f"📋 Created {len(loan_applications)} sample loan applications")
```

```python
# Evaluate all applications with holistic approach
async def evaluate_holistic_decisions():
    results = []
    
    for loan_data in loan_applications:
        name = loan_data['name']
        app = loan_data['app']
        
        print(f"\n🔍 Evaluating: {name}")
        print("=" * 50)
        
        # Base decision
        base_decision = base_engine.evaluate_application(app)
        
        # Reputation analysis (using sample data for demo)
        reputation_score = 85.0 if 'Contributor' in name else 60.0 if 'Average' in name else 35.0
        
        # Adjust decision based on reputation
        final_risk = base_decision.risk_score
        if reputation_score > 80:
            final_risk = max(0, final_risk - 15)
            adjustment = "Reduced risk due to excellent reputation"
        elif reputation_score < 40:
            final_risk = min(100, final_risk + 20)
            adjustment = "Increased risk due to poor reputation"
        else:
            adjustment = "Standard risk assessment"
        
        # Determine final decision
        if final_risk > 70:
            final_decision = "REJECTED"
        elif final_risk > 50:
            final_decision = "NEEDS_REVIEW"
        else:
            final_decision = "APPROVED"
        
        result = {
            'name': name,
            'base_decision': base_decision.decision.value,
            'base_risk': base_decision.risk_score,
            'reputation_score': reputation_score,
            'final_risk': final_risk,
            'final_decision': final_decision,
            'adjustment': adjustment,
            'loan_amount': app.amount,
            'collateral': base_decision.required_collateral
        }
        
        results.append(result)
        
        print(f"📊 Base Decision: {base_decision.decision.value}")
        print(f"🏛️ Reputation Score: {reputation_score:.1f}")
        print(f"🎯 Final Decision: {final_decision}")
        print(f"💡 Adjustment: {adjustment}")
    
    return results

decision_results = await evaluate_holistic_decisions()
```

```python
# Visualize loan decision comparison
df = pd.DataFrame(decision_results)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Decision comparison
decisions = df[['name', 'base_decision', 'final_decision']]
decisions_melted = decisions.melt(id_vars=['name'], var_name='Decision Type', value_name='Decision')

decision_counts = decisions_melted.groupby(['Decision Type', 'Decision']).size().unstack(fill_value=0)
decision_counts.plot(kind='bar', ax=ax1, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
ax1.set_title('Base vs Final Loan Decisions', fontsize=14, fontweight='bold')
ax1.set_ylabel('Number of Applications')
ax1.legend(title='Decision')
ax1.tick_params(axis='x', rotation=45)

# Risk score comparison
x = range(len(df))
width = 0.35
ax2.bar([i - width/2 for i in x], df['base_risk'], width, label='Base Risk', color='lightcoral')
ax2.bar([i + width/2 for i in x], df['final_risk'], width, label='Final Risk', color='lightblue')
ax2.set_title('Risk Score Comparison', fontsize=14, fontweight='bold')
ax2.set_ylabel('Risk Score')
ax2.set_xticks(x)
ax2.set_xticklabels([name.split(' - ')[0] for name in df['name']])
ax2.legend()

# Reputation impact
ax3.scatter(df['reputation_score'], df['final_risk'], s=100, alpha=0.7, color='purple')
for i, name in enumerate(df['name']):
    ax3.annotate(name.split(' - ')[0], (df['reputation_score'].iloc[i], df['final_risk'].iloc[i]), 
                xytext=(5, 5), textcoords='offset points', fontsize=10)
ax3.set_title('Reputation vs Final Risk Score', fontsize=14, fontweight='bold')
ax3.set_xlabel('Reputation Score')
ax3.set_ylabel('Final Risk Score')
ax3.grid(True, alpha=0.3)

# Loan amounts and collateral
ax4.bar(df['name'].str.split(' - ').str[0], df['loan_amount'], alpha=0.7, label='Loan Amount', color='gold')
ax4.bar(df['name'].str.split(' - ').str[0], df['collateral'], alpha=0.7, label='Required Collateral', color='silver')
ax4.set_title('Loan Amount vs Required Collateral', fontsize=14, fontweight='bold')
ax4.set_ylabel('Amount ($)')
ax4.legend()
ax4.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print("📈 Holistic decision analysis visualization complete!")
```
