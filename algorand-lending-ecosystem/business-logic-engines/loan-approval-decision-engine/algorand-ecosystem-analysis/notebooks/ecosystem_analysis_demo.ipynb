## Loan Decision Simulation

Simulate loan approval decisions using the comprehensive ecosystem analysis.

```python
# Simulate loan decisions for different wallet profiles
loan_scenarios = [
    {'profile': 'sophisticated_defi_user', 'amount': 50000, 'duration': 365},
    {'profile': 'governance_participant', 'amount': 25000, 'duration': 180},
    {'profile': 'nft_collector', 'amount': 15000, 'duration': 90},
    {'profile': 'new_user', 'amount': 5000, 'duration': 30},
    {'profile': 'basic_holder', 'amount': 10000, 'duration': 60}
]

loan_decisions = []

for scenario in loan_scenarios:
    profile_name = scenario['profile']
    wallet_address = demo_wallets[profile_name]
    
    print(f"\nProcessing loan request for {profile_name}...")
    
    # Get comprehensive analysis
    analysis_result = await ecosystem_engine.analyze_complete_ecosystem(
        wallet_address,
        analysis_depth="comprehensive",
        requested_loan_amount=scenario['amount']
    )
    
    # Create loan request
    loan_request = LoanRequest(
        borrower_address=wallet_address,
        requested_amount=scenario['amount'],
        loan_purpose="DeFi investment",
        duration_days=scenario['duration'],
        proposed_collateral_amount=scenario['amount'] * 1.5,
        proposed_collateral_assets=["0", "31566704"]  # ALGO and USDC
    )
    
    # Prepare ecosystem data for decision engine
    ecosystem_data = {
        'ecosystem_footprint': analysis_result.ecosystem_footprint,
        'defi_behavior': analysis_result.defi_behavior,
        'governance_participation': analysis_result.governance_participation
    }
    
    # Make loan decision
    loan_decision = await decision_engine.evaluate_loan_request(
        loan_request,
        analysis_result.borrower_profile,
        analysis_result.risk_profile,
        ecosystem_data
    )
    
    loan_decisions.append({
        'profile': profile_name,
        'requested_amount': scenario['amount'],
        'decision': loan_decision.decision_type.value,
        'approved_amount': loan_decision.approved_amount,
        'interest_rate': loan_decision.interest_rate,
        'risk_score': analysis_result.risk_profile.overall_risk_score,
        'confidence': loan_decision.decision_confidence,
        'primary_factors': loan_decision.primary_decision_factors[:3]
    })
    
    print(f"  Decision: {loan_decision.decision_type.value}")
    print(f"  Approved Amount: ${loan_decision.approved_amount:,.2f}")
    print(f"  Interest Rate: {loan_decision.interest_rate:.2%}")
    print(f"  Confidence: {loan_decision.decision_confidence:.1%}")

print("\nLoan decision simulation complete!")
```

## Risk Factor Analysis

Deep dive into risk factors and their impact on loan decisions.

```python
# Analyze risk factors for the sophisticated DeFi user
sophisticated_analysis = comprehensive_result

print("=== Risk Factor Deep Dive ===")
print(f"Wallet: {sophisticated_analysis.borrower_profile.wallet_address[:20]}...")
print(f"\nRisk Profile Breakdown:")

risk_components = {
    'Ecosystem Trust': sophisticated_analysis.risk_profile.ecosystem_trust_score,
    'Financial Sophistication': sophisticated_analysis.risk_profile.financial_sophistication_score,
    'Long-term Commitment': sophisticated_analysis.risk_profile.long_term_commitment_score,
    'Community Standing': sophisticated_analysis.risk_profile.community_standing_score,
    'Volatility Exposure': sophisticated_analysis.risk_profile.volatility_exposure_score,
    'Concentration Risk': sophisticated_analysis.risk_profile.concentration_risk_score,
    'Counterparty Risk': sophisticated_analysis.risk_profile.counterparty_risk_score,
    'Technical Risk': sophisticated_analysis.risk_profile.technical_risk_score
}

# Create risk factor visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Risk component radar chart
import numpy as np

categories = list(risk_components.keys())
values = list(risk_components.values())

# Number of variables
N = len(categories)

# Angle for each category
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]  # Complete the circle

# Values
values += values[:1]  # Complete the circle

# Plot
ax1 = plt.subplot(121, projection='polar')
ax1.plot(angles, values, 'o-', linewidth=2, label='Risk Scores')
ax1.fill(angles, values, alpha=0.25)
ax1.set_xticks(angles[:-1])
ax1.set_xticklabels(categories, fontsize=10)
ax1.set_ylim(0, 100)
ax1.set_title('Risk Component Analysis', size=14, fontweight='bold', pad=20)
ax1.grid(True)

# Risk vs Sophistication scatter
ax2 = plt.subplot(122)
sophistication_factors = {
    'DeFi Protocols': len(sophisticated_analysis.ecosystem_footprint.defi_protocols_used),
    'Governance Votes': sophisticated_analysis.governance_participation.votes_cast,
    'Smart Contracts': sophisticated_analysis.borrower_profile.smart_contract_interactions,
    'Asset Diversity': sophisticated_analysis.ecosystem_footprint.asset_diversity_score * 100,
    'Transaction Volume': min(sophisticated_analysis.borrower_profile.total_volume_algo / 1000, 100)
}

x_vals = list(sophistication_factors.values())
y_vals = [100 - risk_components[factor] for factor in ['Technical Risk', 'Counterparty Risk', 
                                                       'Concentration Risk', 'Volatility Exposure', 
                                                       'Technical Risk']]  # Invert risk for better visualization

colors = ['red', 'orange', 'yellow', 'lightgreen', 'green']
ax2.scatter(x_vals, y_vals, s=100, c=colors, alpha=0.7)

for i, (factor, value) in enumerate(sophistication_factors.items()):
    ax2.annotate(factor, (x_vals[i], y_vals[i]), xytext=(5, 5), 
                textcoords='offset points', fontsize=9)

ax2.set_xlabel('Sophistication Metric Value')
ax2.set_ylabel('Risk Mitigation Score')
ax2.set_title('Sophistication vs Risk Mitigation')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print risk factor details
for factor, score in risk_components.items():
    risk_level = "Low" if score < 30 else "Medium" if score < 70 else "High"
    print(f"  {factor}: {score:.1f}/100 ({risk_level})")

print(f"\nOverall Risk Score: {sophisticated_analysis.risk_profile.overall_risk_score:.1f}/100")
print(f"Risk Level: {sophisticated_analysis.risk_profile.overall_risk_level.value.title()}")
print(f"Confidence Level: {sophisticated_analysis.risk_profile.confidence_level:.1%}")
```

## Pattern Analysis Insights

Analyze detected behavioral patterns and their implications.

```python
# Analyze detected patterns
patterns = sophisticated_analysis.detected_patterns

print("=== Behavioral Pattern Analysis ===")
print(f"Total Patterns Detected: {len(patterns)}\n")

if patterns:
    # Categorize patterns
    positive_patterns = [p for p in patterns if not p.get('risk_indicator', True)]
    risk_patterns = [p for p in patterns if p.get('risk_indicator', False)]
    
    print(f"Positive Patterns: {len(positive_patterns)}")
    for pattern in positive_patterns:
        print(f"  ✓ {pattern['pattern_type']}: {pattern['description']}")
        print(f"    Confidence: {pattern['confidence']:.1%}")
    
    print(f"\nRisk Patterns: {len(risk_patterns)}")
    for pattern in risk_patterns:
        print(f"  ⚠ {pattern['pattern_type']}: {pattern['description']}")
        print(f"    Confidence: {pattern['confidence']:.1%}")
    
    # Pattern confidence distribution
    confidences = [p['confidence'] for p in patterns]
    pattern_types = [p['pattern_type'] for p in patterns]
    
    # Visualization
    plt.figure(figsize=(12, 6))
    
    # Pattern confidence chart
    plt.subplot(1, 2, 1)
    bars = plt.bar(range(len(patterns)), confidences, 
                   color=['green' if not p.get('risk_indicator', True) else 'red' for p in patterns])
    plt.xlabel('Pattern Index')
    plt.ylabel('Confidence Level')
    plt.title('Pattern Detection Confidence')
    plt.xticks(range(len(patterns)), [f"P{i+1}" for i in range(len(patterns))])
    plt.grid(True, alpha=0.3)
    
    # Add pattern type labels
    for i, (bar, pattern_type) in enumerate(zip(bars, pattern_types)):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
                pattern_type.replace('_', '\n'), ha='center', va='bottom', 
                fontsize=8, rotation=45)
    
    # Pattern type distribution
    plt.subplot(1, 2, 2)
    risk_count = len(risk_patterns)
    positive_count = len(positive_patterns)
    
    plt.pie([positive_count, risk_count], 
            labels=['Positive Patterns', 'Risk Patterns'],
            colors=['lightgreen', 'lightcoral'],
            autopct='%1.1f%%', startangle=90)
    plt.title('Pattern Type Distribution')
    
    plt.tight_layout()
    plt.show()
    
else:
    print("No specific patterns detected in this analysis.")
    print("This could indicate either very standard behavior or insufficient data.")

# Ecosystem maturity assessment
maturity = ecosystem_engine._calculate_ecosystem_maturity(sophisticated_analysis)
print(f"\nEcosystem Maturity Assessment: {maturity.title()}")

maturity_factors = {
    'Wallet Age': 'Excellent' if sophisticated_analysis.borrower_profile.wallet_age_days > 730 else 'Good',
    'Transaction Activity': 'High' if sophisticated_analysis.borrower_profile.total_transaction_count > 500 else 'Medium',
    'Governance': 'Active' if sophisticated_analysis.borrower_profile.governance_participation else 'Inactive',
    'DeFi Usage': 'Sophisticated' if len(sophisticated_analysis.ecosystem_footprint.defi_protocols_used) > 3 else 'Basic'
}

print("\nMaturity Factors:")
for factor, level in maturity_factors.items():
    print(f"  {factor}: {level}")
```
