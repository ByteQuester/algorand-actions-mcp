# DeFi Behavior Assessment Engine Demo

This notebook demonstrates the comprehensive DeFi behavior assessment capabilities for loan approval decisions.

## Features Demonstrated:
- Cross-protocol DeFi activity analysis
- Behavioral pattern recognition
- Liquidity behavior assessment
- Risk tolerance evaluation
- Loan approval recommendations


```python
# Setup and imports
import sys
import asyncio
import json
from pathlib import Path
from datetime import datetime, timedelta
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add the core module to the path
sys.path.append(str(Path().parent / "core"))

from core.behavior_engine import BehaviorEngine
from core.protocol_analyzer import ProtocolAnalyzer
from core.behavior_patterns import BehaviorPatternAnalyzer
from core.liquidity_behavior import LiquidityBehaviorAnalyzer

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("🚀 DeFi Behavior Assessment Engine Demo")
print("=" * 50)
```

## 1. Initialize the Behavior Engine

First, let's initialize our comprehensive behavior assessment engine.

```python
# Initialize the behavior engine
behavior_engine = BehaviorEngine()

# Sample addresses for demonstration
sample_addresses = [
    "SOPHISTICATED_ALGO_USER_123456789012345678901234567890",
    "NOVICE_DEFI_USER_123456789012345678901234567890ABCD",
    "HIGH_RISK_TRADER_123456789012345678901234567890EFGH"
]

print("✅ Behavior engine initialized")
print(f"📊 Ready to analyze {len(sample_addresses)} sample addresses")
```

## 2. Comprehensive Behavioral Assessment

Let's run a comprehensive behavioral assessment for our sample addresses.

```python
async def run_behavioral_assessments():
    """Run behavioral assessments for sample addresses"""
    assessments = {}
    
    for i, address in enumerate(sample_addresses):
        print(f"\n🔍 Analyzing address {i+1}: {address[:20]}...")
        
        try:
            # Run comprehensive assessment
            assessment = await behavior_engine.assess_behavioral_risk(
                address=address,
                analysis_period_days=365
            )
            
            assessments[f"User_{i+1}"] = assessment
            
            print(f"   ✅ Assessment completed")
            print(f"   Overall Score: {assessment.overall_behavior_score:.3f}")
            print(f"   Risk Level: {assessment.behavioral_risk_level}")
            print(f"   Recommendation: {assessment.loan_recommendation}")
            
        except Exception as e:
            print(f"   ❌ Assessment failed: {str(e)}")
    
    return assessments

# Run the assessments
assessments = await run_behavioral_assessments()

print(f"\n📈 Completed {len(assessments)} behavioral assessments")
```

## 3. Behavioral Score Analysis

Let's analyze and visualize the behavioral scores across different dimensions.

```python
# Extract scores for analysis
score_data = []

for user_id, assessment in assessments.items():
    score_data.append({
        'User': user_id,
        'Overall Behavior': assessment.overall_behavior_score,
        'Sophistication': assessment.sophistication_score,
        'Risk Management': assessment.risk_management_score,
        'DeFi Experience': assessment.defi_experience_score,
        'Risk Level': assessment.behavioral_risk_level,
        'Recommendation': assessment.loan_recommendation,
        'Rate Adjustment': assessment.interest_rate_adjustment
    })

df_scores = pd.DataFrame(score_data)

# Display summary table
print("📊 Behavioral Assessment Summary:")
print("=" * 50)
print(df_scores.round(3))

# Create visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('DeFi Behavioral Assessment Results', fontsize=16, fontweight='bold')

# 1. Score comparison radar chart
score_columns = ['Overall Behavior', 'Sophistication', 'Risk Management', 'DeFi Experience']
ax1 = axes[0, 0]
df_scores[score_columns].T.plot(kind='bar', ax=ax1)
ax1.set_title('Behavioral Scores by User')
ax1.set_ylabel('Score (0-1)')
ax1.legend(df_scores['User'].tolist())
ax1.tick_params(axis='x', rotation=45)

# 2. Risk level distribution
ax2 = axes[0, 1]
risk_counts = df_scores['Risk Level'].value_counts()
risk_counts.plot(kind='pie', ax=ax2, autopct='%1.1f%%')
ax2.set_title('Risk Level Distribution')
ax2.set_ylabel('')

# 3. Recommendation distribution
ax3 = axes[1, 0]
rec_counts = df_scores['Recommendation'].value_counts()
rec_counts.plot(kind='bar', ax=ax3, color='skyblue')
ax3.set_title('Loan Recommendations')
ax3.set_ylabel('Count')
ax3.tick_params(axis='x', rotation=45)

# 4. Interest rate adjustments
ax4 = axes[1, 1]
df_scores.plot(x='User', y='Rate Adjustment', kind='bar', ax=ax4, color='orange')
ax4.set_title('Interest Rate Adjustments (bps)')
ax4.set_ylabel('Basis Points')
ax4.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 4. Protocol Activity Analysis

Let's dive deeper into cross-protocol activity patterns.

```python
# Analyze protocol activities
protocol_data = []

for user_id, assessment in assessments.items():
    protocol_analysis = assessment.protocol_analysis
    
    protocol_info = {
        'User': user_id,
        'Protocols Used': len(protocol_analysis.protocol_activities),
        'Diversification Score': protocol_analysis.diversification_score,
        'Sophistication Score': protocol_analysis.sophistication_score,
        'Engagement Level': protocol_analysis.engagement_level,
        'Cross-Protocol Strategies': len(protocol_analysis.cross_protocol_strategies)
    }
    
    protocol_data.append(protocol_info)

df_protocols = pd.DataFrame(protocol_data)

print("🔗 Cross-Protocol Activity Analysis:")
print("=" * 40)
print(df_protocols.round(3))

# Visualize protocol engagement
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Cross-Protocol DeFi Engagement Analysis', fontsize=14, fontweight='bold')

# Protocol diversity
ax1 = axes[0]
df_protocols.plot(x='User', y='Protocols Used', kind='bar', ax=ax1, color='lightgreen')
ax1.set_title('Number of Protocols Used')
ax1.set_ylabel('Protocol Count')
ax1.tick_params(axis='x', rotation=45)

# Sophistication vs Diversification
ax2 = axes[1]
scatter = ax2.scatter(df_protocols['Diversification Score'], 
                     df_protocols['Sophistication Score'],
                     s=df_protocols['Protocols Used']*50,
                     alpha=0.6, c=range(len(df_protocols)), cmap='viridis')
ax2.set_xlabel('Diversification Score')
ax2.set_ylabel('Sophistication Score')
ax2.set_title('Sophistication vs Diversification\n(Bubble size = Protocol count)')

# Add user labels
for i, row in df_protocols.iterrows():
    ax2.annotate(row['User'], 
                (row['Diversification Score'], row['Sophistication Score']),
                xytext=(5, 5), textcoords='offset points', fontsize=8)

# Cross-protocol strategies
ax3 = axes[2]
df_protocols.plot(x='User', y='Cross-Protocol Strategies', kind='bar', ax=ax3, color='coral')
ax3.set_title('Advanced Strategy Usage')
ax3.set_ylabel('Strategy Count')
ax3.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 5. Risk Management Analysis

Analyze risk management capabilities and behavioral patterns.

```python
# Extract risk management metrics
risk_data = []

for user_id, assessment in assessments.items():
    behavior_metrics = assessment.behavior_pattern_analysis.behavior_metrics
    
    risk_info = {
        'User': user_id,
        'Risk Tolerance': behavior_metrics.risk_tolerance_score,
        'Loss Management': behavior_metrics.loss_management_score,
        'Profit Taking': behavior_metrics.profit_taking_discipline,
        'Market Timing': behavior_metrics.market_timing_ability,
        'Diversification': behavior_metrics.diversification_discipline,
        'Consistency': behavior_metrics.consistency_score,
        'Overall Risk Level': assessment.behavioral_risk_level
    }
    
    risk_data.append(risk_info)

df_risk = pd.DataFrame(risk_data)

print("⚠️  Risk Management Analysis:")
print("=" * 35)
print(df_risk.round(3))

# Create risk management visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Risk Management Behavioral Analysis', fontsize=16, fontweight='bold')

# Risk management heatmap
ax1 = axes[0, 0]
risk_metrics = ['Risk Tolerance', 'Loss Management', 'Profit Taking', 
               'Market Timing', 'Diversification', 'Consistency']
heatmap_data = df_risk[risk_metrics].T
heatmap_data.columns = df_risk['User']
sns.heatmap(heatmap_data, annot=True, cmap='RdYlGn', ax=ax1, 
           cbar_kws={'label': 'Score (0-1)'})
ax1.set_title('Risk Management Capabilities Heatmap')

# Risk profile radar chart
ax2 = axes[0, 1]
categories = risk_metrics
angles = np.linspace(0, 2*np.pi, len(categories), endpoint=False)
angles = np.concatenate((angles, [angles[0]]))  # Complete the circle

ax2 = plt.subplot(2, 2, 2, projection='polar')
colors = ['red', 'blue', 'green']

for i, (_, row) in enumerate(df_risk.iterrows()):
    values = row[risk_metrics].values
    values = np.concatenate((values, [values[0]]))  # Complete the circle
    ax2.plot(angles, values, 'o-', linewidth=2, 
            label=row['User'], color=colors[i % len(colors)])
    ax2.fill(angles, values, alpha=0.25, color=colors[i % len(colors)])

ax2.set_xticks(angles[:-1])
ax2.set_xticklabels(categories)
ax2.set_ylim(0, 1)
ax2.set_title('Risk Management Radar Chart')
ax2.legend(loc='upper right', bbox_to_anchor=(1.3, 1.0))

# Risk vs Sophistication
ax3 = axes[1, 0]
risk_level_map = {'low': 1, 'medium': 2, 'high': 3, 'very_high': 4}
df_risk['Risk Level Numeric'] = df_risk['Overall Risk Level'].map(risk_level_map)
df_scores_with_risk = df_scores.merge(df_risk[['User', 'Risk Level Numeric']], on='User')

scatter = ax3.scatter(df_scores_with_risk['Sophistication'], 
                     df_scores_with_risk['Risk Level Numeric'],
                     s=100, alpha=0.7, c=range(len(df_scores_with_risk)), cmap='plasma')
ax3.set_xlabel('Sophistication Score')
ax3.set_ylabel('Risk Level (1=Low, 4=Very High)')
ax3.set_title('Sophistication vs Risk Level')

# Add user labels
for i, row in df_scores_with_risk.iterrows():
    ax3.annotate(row['User'], 
                (row['Sophistication'], row['Risk Level Numeric']),
                xytext=(5, 5), textcoords='offset points', fontsize=8)

# Behavioral strengths and weaknesses
ax4 = axes[1, 1]
strengths_data = []
weaknesses_data = []

for user_id, assessment in assessments.items():
    strengths_data.append(len(assessment.strengths))
    weaknesses_data.append(len(assessment.weaknesses))

x = np.arange(len(df_risk['User']))
width = 0.35

ax4.bar(x - width/2, strengths_data, width, label='Strengths', color='green', alpha=0.7)
ax4.bar(x + width/2, weaknesses_data, width, label='Weaknesses', color='red', alpha=0.7)

ax4.set_xlabel('Users')
ax4.set_ylabel('Count')
ax4.set_title('Behavioral Strengths vs Weaknesses')
ax4.set_xticks(x)
ax4.set_xticklabels(df_risk['User'])
ax4.legend()

plt.tight_layout()
plt.show()
```

## 6. Liquidity Behavior Assessment

Analyze liquidity provision behavior and sophistication.

```python
# Extract liquidity behavior data
liquidity_data = []

for user_id, assessment in assessments.items():
    liquidity_analysis = assessment.liquidity_behavior_analysis
    
    liquidity_info = {
        'User': user_id,
        'Total Positions': liquidity_analysis.total_positions,
        'Active Positions': liquidity_analysis.active_positions,
        'Total Liquidity': liquidity_analysis.total_liquidity_provided_usd,
        'Avg Duration': liquidity_analysis.average_position_duration,
        'Sophistication': liquidity_analysis.liquidity_sophistication_score,
        'Risk Management': liquidity_analysis.risk_management_score,
        'Overall Score': liquidity_analysis.overall_liquidity_score
    }
    
    liquidity_data.append(liquidity_info)

df_liquidity = pd.DataFrame(liquidity_data)

print("💧 Liquidity Behavior Analysis:")
print("=" * 35)
print(df_liquidity.round(3))

# Visualize liquidity behavior
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Liquidity Provision Behavior Analysis', fontsize=16, fontweight='bold')

# Liquidity provision activity
ax1 = axes[0, 0]
df_liquidity.plot(x='User', y=['Total Positions', 'Active Positions'], 
                 kind='bar', ax=ax1)
ax1.set_title('Liquidity Provision Activity')
ax1.set_ylabel('Position Count')
ax1.tick_params(axis='x', rotation=45)
ax1.legend()

# Total liquidity provided
ax2 = axes[0, 1]
df_liquidity.plot(x='User', y='Total Liquidity', kind='bar', ax=ax2, color='blue')
ax2.set_title('Total Liquidity Provided')
ax2.set_ylabel('USD Value')
ax2.tick_params(axis='x', rotation=45)

# Sophistication vs Risk Management
ax3 = axes[1, 0]
scatter = ax3.scatter(df_liquidity['Sophistication'], 
                     df_liquidity['Risk Management'],
                     s=df_liquidity['Total Positions']*20 + 50,
                     alpha=0.7, c=range(len(df_liquidity)), cmap='viridis')
ax3.set_xlabel('Liquidity Sophistication Score')
ax3.set_ylabel('Risk Management Score')
ax3.set_title('LP Sophistication vs Risk Management\n(Bubble size = Position count)')

# Add user labels
for i, row in df_liquidity.iterrows():
    ax3.annotate(row['User'], 
                (row['Sophistication'], row['Risk Management']),
                xytext=(5, 5), textcoords='offset points', fontsize=8)

# Overall liquidity scores
ax4 = axes[1, 1]
df_liquidity.plot(x='User', y='Overall Score', kind='bar', ax=ax4, color='orange')
ax4.set_title('Overall Liquidity Behavior Score')
ax4.set_ylabel('Score (0-1)')
ax4.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 7. Loan Decision Analysis

Analyze loan approval decisions and interest rate adjustments.

```python
# Analyze loan decisions with different scenarios
loan_scenarios = [
    {'amount': 5000, 'duration': 30, 'purpose': 'trading'},
    {'amount': 15000, 'duration': 90, 'purpose': 'yield_farming'},
    {'amount': 50000, 'duration': 180, 'purpose': 'arbitrage'}
]

async def analyze_loan_scenarios():
    """Analyze how different loan scenarios affect decisions"""
    scenario_results = []
    
    for i, scenario in enumerate(loan_scenarios):
        print(f"\n💰 Analyzing Loan Scenario {i+1}: ${scenario['amount']:,} for {scenario['duration']} days")
        
        for j, address in enumerate(sample_addresses[:2]):  # Test with first 2 addresses
            try:
                assessment = await behavior_engine.assess_behavioral_risk(
                    address=address,
                    analysis_period_days=365,
                    loan_request=scenario
                )
                
                result = {
                    'Scenario': f"Scenario {i+1}",
                    'User': f"User {j+1}",
                    'Loan Amount': scenario['amount'],
                    'Duration': scenario['duration'],
                    'Purpose': scenario['purpose'],
                    'Behavior Score': assessment.overall_behavior_score,
                    'Risk Level': assessment.behavioral_risk_level,
                    'Recommendation': assessment.loan_recommendation,
                    'Rate Adjustment': assessment.interest_rate_adjustment
                }
                
                scenario_results.append(result)
                
                print(f"   User {j+1}: {assessment.loan_recommendation} ({assessment.interest_rate_adjustment:+.0f} bps)")
                
            except Exception as e:
                print(f"   User {j+1}: Error - {str(e)}")
    
    return scenario_results

# Run scenario analysis
scenario_results = await analyze_loan_scenarios()
df_scenarios = pd.DataFrame(scenario_results)

print("\n📊 Loan Scenario Analysis Results:")
print("=" * 40)
if not df_scenarios.empty:
    print(df_scenarios[['Scenario', 'User', 'Loan Amount', 'Recommendation', 'Rate Adjustment']].to_string(index=False))
else:
    print("No scenario results available")
```

```python
# Visualize loan decision patterns
if not df_scenarios.empty:
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    fig.suptitle('Loan Decision Analysis Across Scenarios', fontsize=14, fontweight='bold')
    
    # Rate adjustments by scenario
    ax1 = axes[0]
    pivot_rate = df_scenarios.pivot(index='User', columns='Scenario', values='Rate Adjustment')
    pivot_rate.plot(kind='bar', ax=ax1)
    ax1.set_title('Interest Rate Adjustments by Scenario')
    ax1.set_ylabel('Basis Points')
    ax1.tick_params(axis='x', rotation=45)
    ax1.legend(title='Scenario')
    
    # Recommendation distribution
    ax2 = axes[1]
    rec_counts = df_scenarios['Recommendation'].value_counts()
    rec_counts.plot(kind='pie', ax=ax2, autopct='%1.1f%%')
    ax2.set_title('Loan Recommendation Distribution')
    ax2.set_ylabel('')
    
    # Loan amount vs rate adjustment
    ax3 = axes[2]
    scatter = ax3.scatter(df_scenarios['Loan Amount'], 
                         df_scenarios['Rate Adjustment'],
                         c=df_scenarios['Behavior Score'], 
                         cmap='RdYlGn', s=100, alpha=0.7)
    ax3.set_xlabel('Loan Amount ($)')
    ax3.set_ylabel('Rate Adjustment (bps)')
    ax3.set_title('Loan Amount vs Rate Adjustment\n(Color = Behavior Score)')
    plt.colorbar(scatter, ax=ax3, label='Behavior Score')
    
    plt.tight_layout()
    plt.show()
else:
    print("⚠️  No scenario data available for visualization")
```
