## Sample Wallet Analysis

Let's analyze the reputation of sample Algorand addresses.

```python
# Sample Algorand addresses for demonstration
# Note: These are example addresses - replace with real addresses for actual analysis
sample_addresses = [
    "ALGORAND_SAMPLE_ADDRESS_1_FOR_DEMO_PURPOSES_TESTING",
    "ALGORAND_SAMPLE_ADDRESS_2_FOR_DEMO_PURPOSES_TESTING",
    "ALGORAND_SAMPLE_ADDRESS_3_FOR_DEMO_PURPOSES_TESTING"
]

print(f"📋 Analyzing {len(sample_addresses)} sample addresses...")
print("Note: In a real implementation, these would be actual Algorand addresses")
```

```python
# Mock analysis function for demonstration
async def mock_reputation_analysis(address):
    """Mock reputation analysis for demo purposes"""
    import random
    
    # Simulate different reputation profiles
    profiles = {
        "high_rep": {
            "transaction_score": random.uniform(0.8, 0.95),
            "balance_score": random.uniform(0.75, 0.9),
            "asset_score": random.uniform(0.7, 0.85),
            "defi_score": random.uniform(0.8, 0.95),
            "governance_score": random.uniform(0.85, 0.95)
        },
        "medium_rep": {
            "transaction_score": random.uniform(0.5, 0.7),
            "balance_score": random.uniform(0.4, 0.6),
            "asset_score": random.uniform(0.3, 0.5),
            "defi_score": random.uniform(0.4, 0.6),
            "governance_score": random.uniform(0.3, 0.5)
        },
        "low_rep": {
            "transaction_score": random.uniform(0.2, 0.4),
            "balance_score": random.uniform(0.1, 0.3),
            "asset_score": random.uniform(0.1, 0.2),
            "defi_score": random.uniform(0.1, 0.3),
            "governance_score": random.uniform(0.1, 0.3)
        }
    }
    
    # Randomly assign a profile
    profile_type = random.choice(["high_rep", "medium_rep", "low_rep"])
    profile = profiles[profile_type]
    
    # Calculate overall score
    wallet_score = (
        profile["transaction_score"] * 0.25 +
        profile["balance_score"] * 0.20 +
        profile["asset_score"] * 0.15 +
        profile["defi_score"] * 0.25 +
        0.15 * 0.5  # longevity placeholder
    )
    
    combined_score = wallet_score * 0.6 + profile["governance_score"] * 0.4
    
    # Determine risk tier
    if combined_score >= 0.85:
        tier, adjustment = "excellent", -0.5
    elif combined_score >= 0.70:
        tier, adjustment = "good", -0.25
    elif combined_score >= 0.50:
        tier, adjustment = "average", 0.0
    elif combined_score >= 0.30:
        tier, adjustment = "poor", 0.25
    else:
        tier, adjustment = "very_poor", 0.5
    
    return {
        'address': address,
        'overall_score': combined_score,
        'risk_tier': tier,
        'rate_adjustment': adjustment,
        'wallet_score': wallet_score,
        'governance_score': profile["governance_score"],
        'transaction_score': profile["transaction_score"],
        'balance_score': profile["balance_score"],
        'asset_score': profile["asset_score"],
        'defi_score': profile["defi_score"],
        'profile_type': profile_type
    }

# Run analysis for all sample addresses
async def analyze_all_addresses():
    results = []
    for address in sample_addresses:
        result = await mock_reputation_analysis(address)
        results.append(result)
        print(f"✅ Analyzed {address[:20]}...")
    return results

# Execute the analysis
analysis_results = await analyze_all_addresses()
print(f"\n🎉 Completed analysis of {len(analysis_results)} addresses")
```

## Recommendations

Based on the analysis, let's generate recommendations for interest rate adjustments.

```python
# Generate recommendations
def generate_recommendations(df_results, portfolio_stats):
    recommendations = []
    
    # Portfolio-level recommendations
    if portfolio_stats['portfolio_average'] > 0.8:
        recommendations.append("✅ Portfolio shows excellent reputation - consider preferential lending terms")
    elif portfolio_stats['portfolio_average'] < 0.4:
        recommendations.append("⚠️ Portfolio shows poor reputation - implement enhanced risk monitoring")
    
    # Rate adjustment recommendations
    if portfolio_stats['avg_rate_adjustment'] < -0.1:
        recommendations.append("💰 Portfolio qualifies for reduced interest rates")
    elif portfolio_stats['avg_rate_adjustment'] > 0.1:
        recommendations.append("📈 Portfolio requires increased interest rates due to risk")
    
    # Individual address recommendations
    excellent_addresses = df_results[df_results['risk_tier'] == 'excellent']
    if len(excellent_addresses) > 0:
        recommendations.append(f"⭐ {len(excellent_addresses)} addresses qualify for premium lending terms")
    
    high_risk_addresses = df_results[df_results['risk_tier'].isin(['poor', 'very_poor'])]
    if len(high_risk_addresses) > 0:
        recommendations.append(f"🔍 {len(high_risk_addresses)} addresses require additional due diligence")
    
    # Governance participation recommendations
    low_governance = df_results[df_results['governance_score'] < 0.3]
    if len(low_governance) > 0:
        recommendations.append(f"🗳️ {len(low_governance)} addresses show low governance participation - encourage community engagement")
    
    return recommendations

recommendations = generate_recommendations(df_results, portfolio_stats)

print("💡 Lending Recommendations")
print("=" * 40)
for i, rec in enumerate(recommendations, 1):
    print(f"{i}. {rec}")

if not recommendations:
    print("📋 No specific recommendations - portfolio shows average risk profile")
```

## Rate Adjustment Calculator

Let's create a simple calculator for determining interest rate adjustments based on reputation scores.

```python
def calculate_adjusted_rate(base_rate, reputation_score, tier_adjustments):
    """Calculate adjusted interest rate based on reputation"""
    
    # Determine tier
    if reputation_score >= 0.85:
        tier = "excellent"
    elif reputation_score >= 0.70:
        tier = "good"
    elif reputation_score >= 0.50:
        tier = "average"
    elif reputation_score >= 0.30:
        tier = "poor"
    else:
        tier = "very_poor"
    
    adjustment = tier_adjustments[tier]
    adjusted_rate = base_rate + adjustment
    
    return adjusted_rate, tier, adjustment

# Example calculation
base_interest_rate = 5.0  # 5% base rate
tier_adjustments = {
    'excellent': -0.5,
    'good': -0.25,
    'average': 0.0,
    'poor': 0.25,
    'very_poor': 0.5
}

print("🧮 Interest Rate Calculator")
print("=" * 40)
print(f"Base Interest Rate: {base_interest_rate}%")
print("\nAdjusted Rates by Address:")
print("-" * 60)

for i, row in df_results.iterrows():
    adjusted_rate, tier, adjustment = calculate_adjusted_rate(
        base_interest_rate, row['overall_score'], tier_adjustments
    )
    
    address_short = row['address'][:20] + '...'
    print(f"{address_short:<25} | Score: {row['overall_score']:.3f} | Tier: {tier:<10} | Rate: {adjusted_rate:.2f}% ({adjustment:+.2f}%)")

# Calculate portfolio average rate
portfolio_avg_rate = base_interest_rate + portfolio_stats['avg_rate_adjustment']
print(f"\n📊 Portfolio Average Rate: {portfolio_avg_rate:.3f}%")
```
