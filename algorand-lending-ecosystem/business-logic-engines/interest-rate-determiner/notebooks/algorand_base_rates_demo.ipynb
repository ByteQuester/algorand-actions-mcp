# Algorand Base Rates Engine Demo

This notebook demonstrates the Algorand Base Rates Engine, which calculates base interest rates from ALGO staking yields and consensus participation rewards.

## Features Demonstrated:
- ALGO staking yield analysis
- Consensus participation rewards
- Governance rewards calculation
- Network security metrics
- Base rate calculation with components breakdown


```python
# Setup and imports
import sys
import asyncio
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Add the engine path
engine_path = Path.cwd().parent / "algorand-base-rates"
sys.path.insert(0, str(engine_path))

from core.algo_staking_engine import AlgoStakingEngine
from core.consensus_rewards import ConsensusRewardsAnalyzer

print("✅ Imports successful!")
print(f"Engine path: {engine_path}")
```

## 1. Initialize Engines

```python
# Initialize the engines
staking_engine = AlgoStakingEngine()
consensus_analyzer = ConsensusRewardsAnalyzer()

print("Engines initialized successfully!")
```

## 2. Current ALGO Staking Metrics

```python
# Get current staking metrics
metrics = await staking_engine.get_current_staking_metrics()

print("=== Current ALGO Staking Metrics ===")
print(f"Current APY: {metrics.current_apy:.2%}")
print(f"Participation Rate: {metrics.participation_rate:.2%}")
print(f"Total Stake: {metrics.total_stake:,} microALGO")
print(f"Online Stake: {metrics.online_stake:,} microALGO")
print(f"Rewards per Round: {metrics.rewards_per_round:,}")
print(f"Confidence Score: {metrics.confidence_score:.2f}")
print(f"Last Update: {metrics.timestamp}")
```

## 3. Consensus Participation Analysis

```python
# Get consensus metrics
consensus_metrics = await consensus_analyzer.get_consensus_metrics()

print("=== Consensus Participation Metrics ===")
print(f"Online Stake Percentage: {consensus_metrics.online_stake_percentage:.2%}")
print(f"Participation Rewards APY: {consensus_metrics.participation_rewards_apy:.2%}")
print(f"Governance Rewards APY: {consensus_metrics.governance_rewards_apy:.2%}")
print(f"Validator Count: {consensus_metrics.validator_count:,}")
print(f"Average Uptime: {consensus_metrics.average_uptime:.2%}")
print(f"Consensus Efficiency: {consensus_metrics.consensus_efficiency:.2%}")
print(f"Confidence Score: {consensus_metrics.confidence_score:.2f}")
```

## 4. Base Rate Calculation with Component Breakdown

```python
# Calculate base rate
base_rate_result = await staking_engine.calculate_base_rate()

print("=== Base Rate Calculation ===")
print(f"Total Base Rate: {base_rate_result.base_rate:.2%}")
print(f"\nComponent Breakdown:")
print(f"  Staking Component: {base_rate_result.staking_component:.2%}")
print(f"  Participation Component: {base_rate_result.participation_component:.2%}")
print(f"  Governance Component: {base_rate_result.governance_component:.2%}")
print(f"  Network Security Component: {base_rate_result.network_security_component:.2%}")
print(f"\nConfidence Score: {base_rate_result.confidence_score:.2f}")
print(f"Calculation Time: {base_rate_result.timestamp}")

# Metadata
print(f"\nMetadata:")
for key, value in base_rate_result.metadata.items():
    print(f"  {key}: {value}")
```

## 5. Governance Analysis

```python
# Get governance summary
governance = await consensus_analyzer.get_governance_summary()

print("=== Governance Summary ===")
print(f"Current Period: {governance.current_period}")
print(f"Total Governors: {governance.total_governors:,}")
print(f"Committed ALGO: {governance.committed_algo / 1_000_000:,.0f}M ALGO")
print(f"Governance APY: {governance.governance_apy:.2%}")
print(f"Voting Participation Rate: {governance.voting_participation_rate:.2%}")
print(f"Period End Date: {governance.period_end_date}")
```

## 6. Reward Component Analysis

```python
# Analyze reward components
reward_analysis = await consensus_analyzer.analyze_reward_components()

print("=== Reward Component Analysis ===")
print(f"Total Reward Rate: {reward_analysis.total_reward_rate:.2%}")
print(f"\nComponent Breakdown:")
print(f"  Consensus Component: {reward_analysis.consensus_component:.2%}")
print(f"  Governance Component: {reward_analysis.governance_component:.2%}")
print(f"  Security Premium: {reward_analysis.security_premium:.2%}")
print(f"  Participation Bonus: {reward_analysis.participation_bonus:.2%}")
print(f"  Risk Adjustment: {reward_analysis.risk_adjustment:.2%}")

print(f"\nMetadata:")
for key, value in reward_analysis.metadata.items():
    print(f"  {key}: {value}")
```

## 7. Historical Analysis

```python
# Get historical analysis for staking
staking_history = await staking_engine.get_historical_analysis(30)

if "error" not in staking_history:
    print("=== 30-Day Staking Historical Analysis ===")
    print(f"Average Rate: {staking_history['average_rate']:.2%}")
    print(f"Median Rate: {staking_history['median_rate']:.2%}")
    print(f"Min Rate: {staking_history['min_rate']:.2%}")
    print(f"Max Rate: {staking_history['max_rate']:.2%}")
    print(f"Standard Deviation: {staking_history['std_deviation']:.2%}")
    print(f"Trend: {staking_history['trend']}")
    print(f"Volatility: {staking_history['volatility']:.2%}")
    print(f"Data Points: {staking_history['data_points']}")
else:
    print(f"Historical analysis error: {staking_history['error']}")

# Get participation trend
participation_trend = await consensus_analyzer.get_participation_trend(30)

if "error" not in participation_trend:
    print("\n=== 30-Day Participation Trend ===")
    print(f"Average Participation: {participation_trend['average_participation']:.2%}")
    print(f"Current Participation: {participation_trend['current_participation']:.2%}")
    print(f"Min Participation: {participation_trend['min_participation']:.2%}")
    print(f"Max Participation: {participation_trend['max_participation']:.2%}")
    print(f"Trend: {participation_trend['trend']}")
    print(f"Volatility: {participation_trend['volatility']:.2%}")
    print(f"Above Target Days: {participation_trend['above_target_days']}/{participation_trend['period_days']}")
    print(f"Target Achievement Rate: {participation_trend['target_achievement_rate']:.2%}")
else:
    print(f"Participation trend error: {participation_trend['error']}")
```

## 8. Visualization: Rate Components Breakdown

```python
# Create visualization of base rate components
components = {
    'Staking': base_rate_result.staking_component,
    'Participation': base_rate_result.participation_component,
    'Governance': base_rate_result.governance_component,
    'Security': base_rate_result.network_security_component
}

# Create pie chart
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Pie chart of components
labels = list(components.keys())
values = [comp * 100 for comp in components.values()]  # Convert to percentages
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4']

ax1.pie(values, labels=labels, autopct='%1.2f%%', startangle=90, colors=colors)
ax1.set_title(f'Base Rate Components\nTotal: {base_rate_result.base_rate:.2%}', fontsize=14, fontweight='bold')

# Bar chart of components
bars = ax2.bar(labels, values, color=colors, alpha=0.8)
ax2.set_ylabel('Rate (%)', fontweight='bold')
ax2.set_title('Base Rate Components Breakdown', fontsize=14, fontweight='bold')
ax2.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bar, value in zip(bars, values):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.01,
             f'{value:.2f}%', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

print(f"\nTotal Base Rate: {base_rate_result.base_rate:.2%}")
print(f"Confidence Score: {base_rate_result.confidence_score:.2f}")
```
