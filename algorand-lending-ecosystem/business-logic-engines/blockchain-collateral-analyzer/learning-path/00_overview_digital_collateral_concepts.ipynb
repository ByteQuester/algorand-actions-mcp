## Key Risk Factors in Digital Collateral

**📉 Price Volatility**: Crypto prices can drop 20-50% in days, requiring margin calls

**🏃 Liquidity Risk**: Some assets may be hard to sell quickly without price impact

**🔮 Oracle Risk**: Price feeds can fail or be manipulated, affecting liquidations

**⚙️ Smart Contract Risk**: Bugs in automated systems could lock or lose collateral

**🌊 Correlation Risk**: Many crypto assets move together, reducing diversification benefits

**⚡ Liquidation Speed**: Need to liquidate quickly before prices fall further
