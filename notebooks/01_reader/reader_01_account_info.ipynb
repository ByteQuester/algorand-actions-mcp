# Account Information Retrieval

This notebook demonstrates how to retrieve account information from the Algorand blockchain using the Reader MCP service.

```python
import sys
sys.path.append('../00_setup')
from helpers import call_api, check_health, format_algo, short_address
from config import READER_URL, ALICE_ADDRESS, BOB_ADDRESS
```

## Health Check

First, let's check if the Reader MCP service is running:

```python
health = await check_health(READER_URL)
print(f"Reader MCP Service: {'✅ Healthy' if health else '❌ Not available'}")

if not health:
    print("💡 Run './setup.sh' to start services or set USE_MOCK_MODE=True in config.py")
```

## Get Account Information

Let's retrieve information for Alice's account:

```python
# Get Alice's account information
result = await call_api(f"{READER_URL}/tools/get_account_info", {"address": ALICE_ADDRESS})

if result.get("success"):
    account = result["account"]
    print(f"📊 Account Information for {short_address(ALICE_ADDRESS)}")
    print(f"💰 Balance: {format_algo(account['amount'])} ALGO")
    print(f"📍 Status: {account['status']}")
    print(f"🔢 Round: {account['round']}")
    
    if account.get('assets'):
        print(f"🪙 Assets: {len(account['assets'])} assets held")
    else:
        print("🪙 Assets: No assets held")
else:
    print(f"❌ Error: {result.get('error')}")
    if result.get('mock_available'):
        print("💡 Set USE_MOCK_MODE=True in config.py to use mock data")
```

## Compare Multiple Accounts

Let's compare Alice and Bob's accounts:

```python
# Get both accounts
alice_result = await call_api(f"{READER_URL}/tools/get_account_info", {"address": ALICE_ADDRESS})
bob_result = await call_api(f"{READER_URL}/tools/get_account_info", {"address": BOB_ADDRESS})

print("📊 Account Comparison")
print("=" * 50)

if alice_result.get("success"):
    alice = alice_result["account"]
    print(f"👩 Alice ({short_address(ALICE_ADDRESS)}): {format_algo(alice['amount'])} ALGO")
else:
    print(f"👩 Alice: ❌ {alice_result.get('error')}")

if bob_result.get("success"):
    bob = bob_result["account"]
    print(f"👨 Bob ({short_address(BOB_ADDRESS)}): {format_algo(bob['amount'])} ALGO")
else:
    print(f"👨 Bob: ❌ {bob_result.get('error')}")
```

## Account Details Analysis

Let's analyze the detailed account structure:

```python
result = await call_api(f"{READER_URL}/tools/get_account_info", {"address": ALICE_ADDRESS})

if result.get("success"):
    import json
    print("🔍 Full Account Data Structure:")
    print(json.dumps(result, indent=2))
else:
    print(f"❌ Error retrieving account data: {result.get('error')}")
```
