## Mock Asset Transfer

Since the current Writer service focuses on payment transactions, let's demonstrate the concept of asset transfers:

```python
# Mock asset transfer data
asset_transfer = {
    "type": "asset_transfer",
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "assetId": USDC_ASSET_ID,
    "amount": TEN_USDC,  # 10 USDC
    "note": "USDC transfer from Alice to Bob"
}

print("🪙 Asset Transfer Transaction (Mock)")
print("=" * 40)
print(f"💰 Asset: USDC (ID: {USDC_ASSET_ID})")
print(f"📊 Amount: {format_asset(TEN_USDC, 6)} USDC")
print(f"👩 From: {short_address(ALICE_ADDRESS)}")
print(f"👨 To: {short_address(BOB_ADDRESS)}")
print(f"📝 Note: {asset_transfer['note']}")

# This would be the structure for a real asset transfer
print("\n📋 Transaction Structure:")
import json
print(json.dumps(asset_transfer, indent=2))
```

## Asset Transfer Scenarios

Let's explore different asset transfer scenarios:

```python
# Different asset transfer scenarios
scenarios = [
    {
        "name": "Small USDC Transfer",
        "assetId": USDC_ASSET_ID,
        "amount": 1_000_000,  # 1 USDC
        "decimals": 6,
        "symbol": "USDC"
    },
    {
        "name": "Large USDC Transfer",
        "assetId": USDC_ASSET_ID,
        "amount": 100_000_000,  # 100 USDC
        "decimals": 6,
        "symbol": "USDC"
    },
    {
        "name": "Fractional Transfer",
        "assetId": USDC_ASSET_ID,
        "amount": 500_000,  # 0.5 USDC
        "decimals": 6,
        "symbol": "USDC"
    }
]

print("🔄 Asset Transfer Scenarios")
print("=" * 40)

for i, scenario in enumerate(scenarios, 1):
    print(f"\n{i}. {scenario['name']}")
    print(f"   Asset ID: {scenario['assetId']}")
    print(f"   Amount: {format_asset(scenario['amount'], scenario['decimals'])} {scenario['symbol']}")
    print(f"   Raw Amount: {scenario['amount']:,} micro-units")
    
    # Calculate transaction fee (would be determined by network)
    base_fee = 1000  # 0.001 ALGO in microAlgos
    print(f"   Estimated Fee: {base_fee} microAlgos (0.001 ALGO)")
```

## Asset Opt-in Transaction

Before receiving an asset, an account must opt-in. Let's show what an opt-in transaction would look like:

```python
# Asset opt-in transaction structure
opt_in_transaction = {
    "type": "asset_transfer",
    "fromAddress": BOB_ADDRESS,
    "toAddress": BOB_ADDRESS,  # Same address for opt-in
    "assetId": USDC_ASSET_ID,
    "amount": 0,  # Zero amount for opt-in
    "note": "Opt-in to USDC asset"
}

print("🔐 Asset Opt-in Transaction (Mock)")
print("=" * 40)
print(f"🪙 Asset: USDC (ID: {USDC_ASSET_ID})")
print(f"👤 Account: {short_address(BOB_ADDRESS)}")
print(f"💰 Amount: 0 (opt-in transaction)")
print(f"📝 Purpose: Enable account to receive USDC")

print("\n📋 Opt-in Transaction Structure:")
print(json.dumps(opt_in_transaction, indent=2))

print("\n💡 Note: Opt-in transactions allow an account to receive a specific asset.")
print("💡 The amount is 0, but there's still a transaction fee of ~0.001 ALGO.")
```

## Asset Close-out Transaction

An account can also close out of an asset, returning remaining balance to another account:

```python
# Asset close-out transaction
close_out_transaction = {
    "type": "asset_transfer",
    "fromAddress": BOB_ADDRESS,
    "toAddress": ALICE_ADDRESS,  # Remaining balance goes here
    "assetId": USDC_ASSET_ID,
    "amount": 0,  # Amount to transfer (0 for close-out)
    "closeRemainderTo": ALICE_ADDRESS,  # All remaining balance
    "note": "Close out USDC position"
}

print("🔒 Asset Close-out Transaction (Mock)")
print("=" * 40)
print(f"🪙 Asset: USDC (ID: {USDC_ASSET_ID})")
print(f"👤 Closing Account: {short_address(BOB_ADDRESS)}")
print(f"👤 Remainder Recipient: {short_address(ALICE_ADDRESS)}")
print(f"📝 Purpose: Close asset position and transfer remaining balance")

print("\n📋 Close-out Transaction Structure:")
print(json.dumps(close_out_transaction, indent=2))

print("\n💡 Note: Close-out removes the asset from the account's holdings.")
print("💡 Any remaining balance is sent to the specified remainder recipient.")
```
