# Writer MCP: Simple Payment

Learn how to build basic payment transactions.

```python
import sys
sys.path.append('../00_setup')
from config import WRITER_URL, ALICE_ADDRESS, BOB_ADDRESS, ONE_ALGO
from helpers import call_api, format_algo
```

## Build Basic Payment

```python
# Build 1 ALGO payment from Alice to Bob
result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", {
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "microAlgos": ONE_ALGO
})
print(f"Payment build: {'✅ Success' if result.get('success') else '❌ Failed'}")
```

## Add Transaction Note

```python
# Build payment with note
result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", {
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "microAlgos": ONE_ALGO,
    "note": "Payment for services"
})
if result.get('success'):
    print("✅ Payment with note built successfully")
    print(f"Transaction ready for signing")
```

## Different Payment Amounts

```python
# Test different amounts
amounts = [ONE_ALGO, ONE_ALGO * 2, ONE_ALGO // 2]  # 1, 2, 0.5 ALGO

for amount in amounts:
    result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", {
        "fromAddress": ALICE_ADDRESS,
        "toAddress": BOB_ADDRESS,
        "microAlgos": amount
    })
    status = '✅' if result.get('success') else '❌'
    print(f"{status} {format_algo(amount):.6f} ALGO payment")
```

## Get Transaction Details

```python
# Build payment and examine details
result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", {
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "microAlgos": ONE_ALGO
})
if result.get('success'):
    txn_b64 = result.get('unsignedTxnBase64', '')
    print(f"Transaction size: {len(txn_b64)} bytes")
    print(f"Ready for signing and submission")
```

## Minimum Amount Test

```python
# Test minimum transaction amount (0.001 ALGO)
min_amount = 1000  # 0.001 ALGO in microAlgos
result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", {
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "microAlgos": min_amount
})
success = result.get('success', False)
print(f"Minimum payment: {'✅ Allowed' if success else '❌ Rejected'}")
print(f"Amount: {format_algo(min_amount):.6f} ALGO")
```
