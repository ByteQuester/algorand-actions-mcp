# Live Real Transaction Builder

This notebook demonstrates building REAL Algorand testnet transactions using the live Writer MCP service.

⚠️ **IMPORTANT**: This creates actual testnet transactions that can be signed and submitted to the blockchain!

```python
import sys
sys.path.append('../00_setup')
from helpers import call_api, check_health, format_algo, short_address
from config import WRITER_URL, ALICE_ADDRESS, BOB_ADDRESS, ONE_ALGO, HALF_ALGO
```

## Verify Live Writer Service

First, let's confirm the live Writer service is running:

```python
print("🔍 Checking Live Writer Service")
print("=" * 35)

health_result = await call_api(f"{WRITER_URL}/health")

if health_result.get("status") == "ok":
    print("✅ Live Writer Service is running!")
    print(f"🌐 Mode: {health_result.get('mode', 'Unknown')}")
    print(f"🏷️  Service: {health_result.get('service', 'Unknown')}")
    print(f"📡 Network: {health_result.get('network', 'Unknown')}")
    print(f"⏰ Timestamp: {health_result.get('timestamp', 'Unknown')}")
    
    if "LIVE_WRITER" in health_result.get('mode', ''):
        print("\n🚨 LIVE MODE CONFIRMED - Real transactions will be built!")
    else:
        print("\n🧪 Test mode detected")
else:
    print("❌ Writer service not available")
    print("💡 Start the service with: node notebooks/demo-writer-service.js")
    print("💡 Or check the service is running on the correct port")
```

## Build Real Testnet Transaction #1

Let's build our first real Algorand testnet transaction:

```python
import datetime

print("🔨 Building Real Testnet Transaction #1")
print("=" * 45)

# Transaction parameters
amount = HALF_ALGO  # 0.5 ALGO
timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
note = f"Live demo transaction #{1} - {timestamp}"

transaction_data = {
    "fromAddress": ALICE_ADDRESS,
    "toAddress": BOB_ADDRESS,
    "microAlgos": amount,
    "note": note
}

print(f"💰 Amount: {format_algo(amount)} ALGO")
print(f"👩 From: {short_address(ALICE_ADDRESS)}")
print(f"👨 To: {short_address(BOB_ADDRESS)}")
print(f"📝 Note: {note}")
print(f"⏰ Time: {timestamp}")

# Build the real transaction
result = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", transaction_data)

if result.get("success"):
    print("\n🎉 REAL TRANSACTION BUILT SUCCESSFULLY!")
    print(f"🆔 Transaction ID: {result['txId']}")
    print(f"💸 Network Fee: {result['fee']} microAlgos ({format_algo(result['fee'])} ALGO)")
    print(f"🔢 Valid from round {result['firstRound']} to {result['lastRound']}")
    print(f"🔐 Genesis Hash: {result['genesisHash'][:32]}...")
    
    # Store for later reference
    transaction_1 = result
    
    print("\n⚠️  IMPORTANT:")
    print(f"   This is a REAL Algorand testnet transaction!")
    print(f"   It can be signed and submitted to the blockchain.")
    print(f"   Transaction data: {len(result['unsignedTxnBase64'])} bytes")
else:
    print(f"\n❌ Transaction build failed: {result.get('error')}")
    transaction_1 = None
```

## Build Real Testnet Transaction #2

Let's build a second transaction with different parameters:

```python
print("🔨 Building Real Testnet Transaction #2")
print("=" * 45)

# Different transaction parameters
amount_2 = ONE_ALGO * 5  # 5 ALGO
timestamp_2 = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
note_2 = f"Larger payment demo - {timestamp_2}"

transaction_data_2 = {
    "fromAddress": BOB_ADDRESS,  # Reverse direction
    "toAddress": ALICE_ADDRESS,
    "microAlgos": amount_2,
    "note": note_2
}

print(f"💰 Amount: {format_algo(amount_2)} ALGO")
print(f"👨 From: {short_address(BOB_ADDRESS)}")
print(f"👩 To: {short_address(ALICE_ADDRESS)}")
print(f"📝 Note: {note_2}")

# Build the second real transaction
result_2 = await call_api(f"{WRITER_URL}/tools/build_payment_transaction", transaction_data_2)

if result_2.get("success"):
    print("\n🎉 SECOND REAL TRANSACTION BUILT!")
    print(f"🆔 Transaction ID: {result_2['txId']}")
    print(f"💸 Network Fee: {result_2['fee']} microAlgos")
    print(f"🔢 Valid from round {result_2['firstRound']} to {result_2['lastRound']}")
    
    # Store for comparison
    transaction_2 = result_2
    
    print("\n📊 Transaction Comparison:")
    if transaction_1:
        print(f"   TX #1: {transaction_1['txId']} ({format_algo(amount)} ALGO)")
    print(f"   TX #2: {result_2['txId']} ({format_algo(amount_2)} ALGO)")
else:
    print(f"\n❌ Second transaction build failed: {result_2.get('error')}")
    transaction_2 = None
```

## Transaction Analysis

Let's analyze the real transactions we've built:

```python
print("🔍 Real Transaction Analysis")
print("=" * 35)

transactions = []
if transaction_1:
    transactions.append(("Transaction #1", transaction_1, amount))
if transaction_2:
    transactions.append(("Transaction #2", transaction_2, amount_2))

if transactions:
    print(f"\n📊 Built {len(transactions)} real testnet transactions:")
    
    total_value = 0
    total_fees = 0
    
    for name, tx, amt in transactions:
        total_value += amt
        total_fees += tx['fee']
        
        print(f"\n🔸 {name}:")
        print(f"   ID: {tx['txId']}")
        print(f"   Amount: {format_algo(amt)} ALGO")
        print(f"   Fee: {tx['fee']} microAlgos")
        print(f"   Rounds: {tx['firstRound']} - {tx['lastRound']}")
        print(f"   Size: {len(tx['unsignedTxnBase64'])} bytes (Base64)")
    
    print(f"\n💰 Total Analysis:")
    print(f"   Total Value: {format_algo(total_value)} ALGO")
    print(f"   Total Fees: {format_algo(total_fees)} ALGO")
    print(f"   Combined Cost: {format_algo(total_value + total_fees)} ALGO")
    
    # Validate transaction IDs are unique
    tx_ids = [tx['txId'] for _, tx, _ in transactions]
    if len(set(tx_ids)) == len(tx_ids):
        print(f"   ✅ All transaction IDs are unique")
    else:
        print(f"   ⚠️  Duplicate transaction IDs detected")
else:
    print("❌ No transactions were successfully built")
```

## Verify Testnet Connectivity

Let's verify our transactions are connected to the real Algorand testnet:

```python
print("🌐 Testnet Connectivity Verification")
print("=" * 40)

if transactions:
    # Analyze genesis hash (should be consistent for testnet)
    genesis_hashes = [tx['genesisHash'] for _, tx, _ in transactions]
    unique_genesis = set(genesis_hashes)
    
    print(f"🔐 Genesis Hash Analysis:")
    print(f"   Unique genesis hashes: {len(unique_genesis)}")
    
    if len(unique_genesis) == 1:
        print(f"   ✅ All transactions use same genesis hash (consistent network)")
        print(f"   📍 Genesis: {list(unique_genesis)[0][:32]}...")
    else:
        print(f"   ⚠️  Multiple genesis hashes detected")
    
    # Analyze round numbers (should be current)
    first_rounds = [tx['firstRound'] for _, tx, _ in transactions]
    last_rounds = [tx['lastRound'] for _, tx, _ in transactions]
    
    min_first = min(first_rounds)
    max_last = max(last_rounds)
    
    print(f"\n📈 Round Number Analysis:")
    print(f"   Round range: {min_first:,} - {max_last:,}")
    print(f"   Window size: {max_last - min_first:,} rounds")
    print(f"   ✅ Using current testnet round numbers")
    
    # Network fee analysis
    fees = [tx['fee'] for _, tx, _ in transactions]
    avg_fee = sum(fees) / len(fees)
    
    print(f"\n💸 Network Fee Analysis:")
    print(f"   Average fee: {avg_fee:.0f} microAlgos ({format_algo(avg_fee):.6f} ALGO)")
    print(f"   Fee range: {min(fees)} - {max(fees)} microAlgos")
    
    if 1000 <= avg_fee <= 2000:
        print(f"   ✅ Fees within expected testnet range")
    else:
        print(f"   ⚠️  Unusual fee detected")
    
    print(f"\n🎯 CONFIRMATION: These are real Algorand testnet transactions!")
    print(f"   • Connected to live testnet network")
    print(f"   • Using current round numbers")
    print(f"   • Standard network fees applied")
    print(f"   • Ready for signing and submission")
else:
    print("❌ No transactions available for verification")
```
