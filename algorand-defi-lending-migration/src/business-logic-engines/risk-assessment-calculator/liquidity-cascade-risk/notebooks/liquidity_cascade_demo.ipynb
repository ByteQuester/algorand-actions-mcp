```python
# Slippage analysis for different order sizes
print("\n🎯 Slippage Analysis for Large Orders")
print("="*45)

order_sizes = [100_000, 500_000, 1_000_000, 2_500_000, 5_000_000]  # USD amounts
pairs = ['ALGO/USDC', 'GARD/ALGO']

# Simulate slippage calculations
slippage_data = {}

for pair in pairs:
    slippage_data[pair] = []
    base_liquidity = total_algo_usdc if pair == 'ALGO/USDC' else total_gard_algo
    
    for size in order_sizes:
        # Simple slippage model: slippage increases with size relative to liquidity
        size_ratio = size / base_liquidity
        base_slippage = 0.001  # 0.1% base
        
        # Square root price impact model
        slippage_pct = base_slippage + 0.05 * np.sqrt(size_ratio)
        
        # Convert to basis points
        slippage_bps = slippage_pct * 10000
        
        slippage_data[pair].append({
            'order_size': size,
            'slippage_pct': slippage_pct,
            'slippage_bps': slippage_bps,
            'price_impact': slippage_pct * 1.2  # Price impact usually higher than slippage
        })

# Display slippage table
for pair in pairs:
    print(f"\n📈 {pair} Slippage Analysis:")
    print(f"{'Order Size':<12} {'Slippage':<12} {'Price Impact':<15} {'Risk Level':<12}")
    print("-" * 55)
    
    for data in slippage_data[pair]:
        size_str = f"${data['order_size']/1e6:.1f}M"
        slippage_str = f"{data['slippage_bps']:.0f} bps"
        impact_str = f"{data['price_impact']:.2%}"
        
        # Risk level based on slippage
        if data['slippage_bps'] < 50:
            risk = "LOW"
        elif data['slippage_bps'] < 200:
            risk = "MEDIUM"
        elif data['slippage_bps'] < 500:
            risk = "HIGH"
        else:
            risk = "CRITICAL"
        
        print(f"{size_str:<12} {slippage_str:<12} {impact_str:<15} {risk:<12}")
```

## 5. Correlation and Contagion Analysis

Analyze cross-asset correlations and potential contagion effects.

```python
print("🔗 Correlation and Contagion Analysis")
print("="*50)

# Simulate correlation matrix for major assets
assets = ['ALGO', 'USDC', 'USDT', 'GARD', 'BANK', 'OPUL']
n_assets = len(assets)

# Create realistic correlation matrix
np.random.seed(42)  # For reproducible results

# Base correlations
base_correlations = {
    ('ALGO', 'BANK'): 0.65,    # DeFi tokens correlated
    ('ALGO', 'OPUL'): 0.55,
    ('ALGO', 'GARD'): 0.45,    # Lower correlation with stablecoin
    ('ALGO', 'USDC'): 0.15,    # Low correlation with fiat stablecoin
    ('ALGO', 'USDT'): 0.18,
    ('BANK', 'OPUL'): 0.70,    # DeFi tokens highly correlated
    ('BANK', 'GARD'): 0.35,
    ('OPUL', 'GARD'): 0.30,
    ('USDC', 'USDT'): 0.85,    # Stablecoins highly correlated
    ('USDC', 'GARD'): 0.40,    # Moderate correlation between stablecoins
    ('USDT', 'GARD'): 0.38,
}

# Create correlation matrix
correlation_matrix = np.eye(n_assets)

for i, asset1 in enumerate(assets):
    for j, asset2 in enumerate(assets):
        if i != j:
            key = (asset1, asset2) if (asset1, asset2) in base_correlations else (asset2, asset1)
            if key in base_correlations:
                correlation_matrix[i, j] = base_correlations[key]
            else:
                # Default low correlation with some noise
                correlation_matrix[i, j] = np.random.uniform(0.1, 0.3)

# Ensure matrix is symmetric
correlation_matrix = (correlation_matrix + correlation_matrix.T) / 2
np.fill_diagonal(correlation_matrix, 1.0)

correlation_df = pd.DataFrame(correlation_matrix, index=assets, columns=assets)

print("\n📊 Asset Correlation Matrix:")
print(correlation_df.round(3))
```

```python
# Contagion analysis
print("\n🦠 Contagion Risk Analysis")
print("="*35)

# Simulate contagion scenarios
contagion_scenarios = [
    {
        'source': 'ALGO',
        'shock_magnitude': 0.30,  # 30% price shock
        'description': 'Major ALGO sell-off'
    },
    {
        'source': 'GARD',
        'shock_magnitude': 0.10,  # 10% depeg
        'description': 'GARD stablecoin depeg event'
    },
    {
        'source': 'BANK',
        'shock_magnitude': 0.50,  # 50% crash
        'description': 'BANK protocol exploit'
    }
]

contagion_results = []

for scenario in contagion_scenarios:
    source_idx = assets.index(scenario['source'])
    shock = scenario['shock_magnitude']
    
    # Calculate contagion effects based on correlations
    contagion_effects = {}
    
    for i, asset in enumerate(assets):
        if asset != scenario['source']:
            correlation = correlation_matrix[source_idx, i]
            # Contagion effect = correlation * shock * contagion amplifier
            contagion_amplifier = 0.8  # 80% transmission
            effect = correlation * shock * contagion_amplifier
            contagion_effects[asset] = effect
    
    contagion_results.append({
        'scenario': scenario,
        'effects': contagion_effects,
        'total_system_impact': sum(abs(effect) for effect in contagion_effects.values())
    })
    
    print(f"\n🎯 {scenario['description']}:")
    print(f"   Source: {scenario['source']} (-{shock:.1%})")
    print("   Contagion Effects:")
    
    for asset, effect in sorted(contagion_effects.items(), key=lambda x: abs(x[1]), reverse=True):
        if abs(effect) > 0.01:  # Only show significant effects
            print(f"     {asset}: {effect:+.1%}")
    
    system_impact = sum(abs(effect) for effect in contagion_effects.values())
    print(f"   💥 Total System Impact: {system_impact:.2f}")
```

```python
# Visualize correlation and contagion analysis
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Correlation and Contagion Analysis', fontsize=16, fontweight='bold')

# 1. Correlation Heatmap
sns.heatmap(correlation_df, annot=True, cmap='RdYlBu_r', center=0, 
            square=True, ax=ax1, cbar_kws={'label': 'Correlation'})
ax1.set_title('Asset Correlation Matrix')

# 2. Contagion Impact Comparison
scenario_names = [r['scenario']['source'] + ' Event' for r in contagion_results]
system_impacts = [r['total_system_impact'] for r in contagion_results]

bars = ax2.bar(scenario_names, system_impacts, color=['red', 'orange', 'darkred'], alpha=0.7)
ax2.set_ylabel('Total System Impact')
ax2.set_title('Contagion Scenario Comparison')
ax2.grid(True, alpha=0.3)

# Add value labels on bars
for bar, impact in zip(bars, system_impacts):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.01,
             f'{impact:.2f}', ha='center', va='bottom')

# 3. Asset Vulnerability Ranking
# Calculate average vulnerability across scenarios
asset_vulnerabilities = {}
for asset in assets:
    total_vulnerability = 0
    count = 0
    for result in contagion_results:
        if asset in result['effects']:
            total_vulnerability += abs(result['effects'][asset])
            count += 1
    asset_vulnerabilities[asset] = total_vulnerability / count if count > 0 else 0

vulnerable_assets = sorted(asset_vulnerabilities.items(), key=lambda x: x[1], reverse=True)
vuln_assets, vuln_scores = zip(*vulnerable_assets)

colors = plt.cm.Reds(np.linspace(0.4, 1, len(vuln_assets)))
ax3.barh(vuln_assets, vuln_scores, color=colors)
ax3.set_xlabel('Average Vulnerability Score')
ax3.set_title('Asset Vulnerability Ranking')
ax3.grid(True, alpha=0.3)

# 4. Network Connectivity Analysis
# Calculate asset connectivity (sum of absolute correlations)
connectivity_scores = {}
for i, asset in enumerate(assets):
    connectivity = sum(abs(correlation_matrix[i, j]) for j in range(n_assets) if i != j)
    connectivity_scores[asset] = connectivity

connected_assets = sorted(connectivity_scores.items(), key=lambda x: x[1], reverse=True)
conn_assets, conn_scores = zip(*connected_assets)

ax4.pie(conn_scores, labels=conn_assets, autopct='%1.1f%%', startangle=90)
ax4.set_title('Network Connectivity Distribution')

plt.tight_layout()
plt.show()

print("📊 Correlation and contagion analysis visualization complete")
```

```python
print("🏦 Stablecoin Risk Assessment")
print("="*40)

# Simulate stablecoin data
stablecoins = {
    'USDC': {
        'type': 'Fiat-Backed',
        'current_price': 1.0008,
        'target_peg': 1.0000,
        'market_cap': 100_000_000,
        'reserve_ratio': 1.00,
        'stability_mechanisms': ['Reserves', 'Arbitrage'],
        'daily_volume': 5_000_000,
        'depeg_history_30d': [0.0002, 0.0005, 0.0003, 0.0008, 0.0001]
    },
    'USDT': {
        'type': 'Fiat-Backed',
        'current_price': 0.9995,
        'target_peg': 1.0000,
        'market_cap': 50_000_000,
        'reserve_ratio': 0.95,
        'stability_mechanisms': ['Reserves', 'Arbitrage'],
        'daily_volume': 2_000_000,
        'depeg_history_30d': [0.0003, 0.0012, 0.0008, 0.0005, 0.0015]
    },
    'GARD': {
        'type': 'Algorithmic',
        'current_price': 0.9925,
        'target_peg': 1.0000,
        'market_cap': 10_000_000,
        'reserve_ratio': 1.50,
        'stability_mechanisms': ['Algorithmic', 'Arbitrage', 'Elastic Supply'],
        'daily_volume': 500_000,
        'depeg_history_30d': [0.0025, 0.0075, 0.0050, 0.0100, 0.0035]
    }
}

# Calculate stability metrics
for coin, data in stablecoins.items():
    # Current depeg
    data['current_depeg'] = abs(data['current_price'] - data['target_peg'])
    data['current_depeg_bps'] = data['current_depeg'] * 10000
    
    # Stability score (0-1, higher is more stable)
    avg_depeg = np.mean(data['depeg_history_30d'])
    max_depeg = max(data['depeg_history_30d'])
    volatility = np.std(data['depeg_history_30d'])
    
    # Score components
    current_stability = max(0, 1 - data['current_depeg'] / 0.01)  # Penalty for depeg > 1%
    historical_stability = max(0, 1 - avg_depeg / 0.005)  # Penalty for avg depeg > 0.5%
    consistency = max(0, 1 - volatility / 0.003)  # Penalty for high volatility
    reserve_strength = min(1, data['reserve_ratio'])  # Bonus for over-collateralization
    
    data['stability_score'] = (current_stability * 0.3 + historical_stability * 0.3 + 
                              consistency * 0.2 + reserve_strength * 0.2)
    
    # Risk level
    if data['stability_score'] > 0.8:
        data['risk_level'] = 'LOW'
    elif data['stability_score'] > 0.6:
        data['risk_level'] = 'MEDIUM'
    elif data['stability_score'] > 0.4:
        data['risk_level'] = 'HIGH'
    else:
        data['risk_level'] = 'CRITICAL'

# Display stablecoin analysis
print("\n📊 Stablecoin Stability Analysis:")
print(f"{'Coin':<6} {'Type':<15} {'Price':<8} {'Depeg':<8} {'Score':<8} {'Risk':<10} {'Reserve':<8}")
print("-" * 75)

for coin, data in stablecoins.items():
    print(f"{coin:<6} {data['type']:<15} ${data['current_price']:.4f} "
          f"{data['current_depeg_bps']:.0f} bps {data['stability_score']:.2f}    "
          f"{data['risk_level']:<10} {data['reserve_ratio']:.2f}")

print("\n💡 Key Insights:")
for coin, data in stablecoins.items():
    if data['current_depeg_bps'] > 50:
        print(f"  ⚠️  {coin} showing elevated depeg risk: {data['current_depeg_bps']:.0f} bps")
    if data['stability_score'] < 0.7:
        print(f"  🔍 {coin} requires closer monitoring (stability score: {data['stability_score']:.2f})")
    if data['reserve_ratio'] > 1.2:
        print(f"  ✅ {coin} well-collateralized (reserve ratio: {data['reserve_ratio']:.2f})")
```

## 7. Stress Testing and Monte Carlo Analysis

Perform stress testing and Monte Carlo simulations for comprehensive risk assessment.

```python
print("🎲 Stress Testing & Monte Carlo Analysis")
print("="*50)

# Define stress test scenarios
stress_scenarios = {
    'Market Crash': {
        'description': '50% market crash over 24 hours',
        'algo_shock': -0.50,
        'correlation_increase': 0.30,
        'liquidity_reduction': 0.60,
        'duration_hours': 24,
        'probability': 0.05
    },
    'Flash Crash': {
        'description': '20% flash crash in 1 hour',
        'algo_shock': -0.20,
        'correlation_increase': 0.40,
        'liquidity_reduction': 0.80,
        'duration_hours': 1,
        'probability': 0.10
    },
    'Stablecoin Crisis': {
        'description': 'Major stablecoin depegs 10%',
        'algo_shock': -0.15,
        'correlation_increase': 0.50,
        'liquidity_reduction': 0.40,
        'duration_hours': 12,
        'probability': 0.08
    },
    'Liquidity Crisis': {
        'description': 'Liquidity dries up across DEXs',
        'algo_shock': -0.10,
        'correlation_increase': 0.25,
        'liquidity_reduction': 0.90,
        'duration_hours': 6,
        'probability': 0.15
    },
    'Protocol Hack': {
        'description': 'Major lending protocol compromised',
        'algo_shock': -0.30,
        'correlation_increase': 0.35,
        'liquidity_reduction': 0.30,
        'duration_hours': 72,
        'probability': 0.12
    }
}

# Simulate stress test results
portfolio = sample_portfolio
stress_results = {}

print("\n🔥 Stress Test Results:")
print(f"{'Scenario':<18} {'Portfolio Loss':<15} {'Max Drawdown':<15} {'Recovery Time':<15} {'Risk Level':<12}")
print("-" * 85)

for scenario_name, scenario in stress_scenarios.items():
    # Simulate portfolio impact
    algo_weight = portfolio['ALGO']
    gard_weight = portfolio['GARD']
    bank_weight = portfolio['BANK']
    usdc_weight = portfolio['USDC']
    
    # Calculate portfolio impact
    portfolio_loss = (
        algo_weight * scenario['algo_shock'] +
        gard_weight * scenario['algo_shock'] * 0.6 +  # GARD partially follows ALGO
        bank_weight * scenario['algo_shock'] * 0.8 +  # BANK highly correlated
        usdc_weight * 0.02 * (1 if 'Stablecoin' in scenario_name else 0)  # USDC impact only in stablecoin crisis
    )
    
    # Max drawdown includes liquidity impact
    liquidity_penalty = scenario['liquidity_reduction'] * 0.05  # Additional 5% per liquidity reduction
    max_drawdown = abs(portfolio_loss) + liquidity_penalty
    
    # Recovery time estimation
    base_recovery = scenario['duration_hours'] * 2  # Usually 2x the crisis duration
    severity_multiplier = 1 + abs(portfolio_loss)
    recovery_time = base_recovery * severity_multiplier
    
    # Risk level
    if max_drawdown < 0.10:
        risk_level = "LOW"
    elif max_drawdown < 0.25:
        risk_level = "MEDIUM"
    elif max_drawdown < 0.50:
        risk_level = "HIGH"
    else:
        risk_level = "CRITICAL"
    
    stress_results[scenario_name] = {
        'portfolio_loss': portfolio_loss,
        'max_drawdown': max_drawdown,
        'recovery_time_hours': recovery_time,
        'risk_level': risk_level,
        'probability': scenario['probability']
    }
    
    print(f"{scenario_name:<18} {portfolio_loss:>+8.1%}        {max_drawdown:>8.1%}        "
          f"{recovery_time:>8.0f}h        {risk_level:<12}")

print("\n" + "-" * 85)
```

```python
# Monte Carlo simulation
print("\n🎯 Monte Carlo Simulation (10,000 scenarios)")
print("="*50)

# Simulation parameters
num_simulations = 10_000
time_horizon_days = 30

# Asset return parameters (annual)
asset_params = {
    'ALGO': {'expected_return': 0.15, 'volatility': 0.60},
    'USDC': {'expected_return': 0.02, 'volatility': 0.05},
    'GARD': {'expected_return': 0.05, 'volatility': 0.30},
    'BANK': {'expected_return': 0.25, 'volatility': 0.80}
}

# Run Monte Carlo simulation
np.random.seed(42)  # For reproducible results
portfolio_returns = []

for i in range(num_simulations):
    daily_returns = {}
    
    for asset in portfolio.keys():
        params = asset_params[asset]
        # Generate daily returns for the time horizon
        daily_mean = params['expected_return'] / 365
        daily_vol = params['volatility'] / np.sqrt(365)
        
        asset_returns = np.random.normal(daily_mean, daily_vol, time_horizon_days)
        daily_returns[asset] = asset_returns
    
    # Calculate portfolio returns
    portfolio_daily_returns = np.zeros(time_horizon_days)
    for asset, weight in portfolio.items():
        portfolio_daily_returns += daily_returns[asset] * weight
    
    # Calculate total return
    total_return = (1 + portfolio_daily_returns).prod() - 1
    portfolio_returns.append(total_return)

portfolio_returns = np.array(portfolio_returns)

# Calculate risk metrics
var_95 = np.percentile(portfolio_returns, 5)
var_99 = np.percentile(portfolio_returns, 1)
var_999 = np.percentile(portfolio_returns, 0.1)

expected_return = np.mean(portfolio_returns)
volatility = np.std(portfolio_returns)

# Expected Shortfall (Conditional VaR)
es_95 = np.mean(portfolio_returns[portfolio_returns <= var_95])
es_99 = np.mean(portfolio_returns[portfolio_returns <= var_99])

# Probability of loss
prob_loss = (portfolio_returns < 0).mean()
prob_extreme_loss = (portfolio_returns < -0.20).mean()  # >20% loss

print(f"\n📊 Monte Carlo Results ({time_horizon_days}-day horizon):")
print(f"\n📈 Return Statistics:")
print(f"  Expected Return:        {expected_return:>+8.2%}")
print(f"  Volatility:             {volatility:>8.2%}")
print(f"  Best Case (95th %ile):  {np.percentile(portfolio_returns, 95):>+8.2%}")
print(f"  Worst Case (5th %ile):  {np.percentile(portfolio_returns, 5):>+8.2%}")

print(f"\n⚠️  Risk Metrics:")
print(f"  VaR 95%:                {var_95:>+8.2%}")
print(f"  VaR 99%:                {var_99:>+8.2%}")
print(f"  VaR 99.9%:              {var_999:>+8.2%}")
print(f"  Expected Shortfall 95%: {es_95:>+8.2%}")
print(f"  Expected Shortfall 99%: {es_99:>+8.2%}")

print(f"\n🎲 Probability Analysis:")
print(f"  Probability of Loss:    {prob_loss:>8.1%}")
print(f"  Probability of >20% Loss: {prob_extreme_loss:>6.1%}")

# Sharpe ratio (assuming 2% risk-free rate)
risk_free_rate = 0.02 * (time_horizon_days / 365)  # Adjust for time horizon
sharpe_ratio = (expected_return - risk_free_rate) / volatility
print(f"  Sharpe Ratio:           {sharpe_ratio:>8.2f}")
```

```python
# Visualize stress testing and Monte Carlo results
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Stress Testing and Monte Carlo Analysis', fontsize=16, fontweight='bold')

# 1. Stress Test Scenario Comparison
scenario_names = list(stress_results.keys())
max_drawdowns = [stress_results[s]['max_drawdown'] for s in scenario_names]
probabilities = [stress_results[s]['probability'] for s in scenario_names]

# Create bubble chart (size = probability, color = drawdown)
scatter = ax1.scatter(range(len(scenario_names)), max_drawdowns, 
                     s=[p*2000 for p in probabilities], 
                     c=max_drawdowns, cmap='Reds', alpha=0.7)

ax1.set_xlabel('Stress Scenarios')
ax1.set_ylabel('Maximum Drawdown')
ax1.set_title('Stress Test Scenario Analysis\n(bubble size = probability)')
ax1.set_xticks(range(len(scenario_names)))
ax1.set_xticklabels([name.replace(' ', '\n') for name in scenario_names], fontsize=9)
ax1.grid(True, alpha=0.3)

# Add colorbar
plt.colorbar(scatter, ax=ax1, label='Max Drawdown')

# 2. Monte Carlo Return Distribution
ax2.hist(portfolio_returns, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
ax2.axvline(var_95, color='orange', linestyle='--', linewidth=2, label=f'VaR 95%: {var_95:.1%}')
ax2.axvline(var_99, color='red', linestyle='--', linewidth=2, label=f'VaR 99%: {var_99:.1%}')
ax2.axvline(expected_return, color='green', linestyle='-', linewidth=2, label=f'Expected: {expected_return:.1%}')

ax2.set_xlabel('Portfolio Return')
ax2.set_ylabel('Frequency')
ax2.set_title(f'Monte Carlo Return Distribution\n({num_simulations:,} simulations)')
ax2.legend()
ax2.grid(True, alpha=0.3)

# 3. Risk vs Return Comparison
# Compare stress test vs Monte Carlo results
stress_losses = [abs(stress_results[s]['portfolio_loss']) for s in scenario_names]
stress_probs = [stress_results[s]['probability'] for s in scenario_names]

mc_risk_levels = [abs(var_95), abs(var_99), abs(var_999)]
mc_risk_labels = ['VaR 95%', 'VaR 99%', 'VaR 99.9%']
mc_probs = [0.05, 0.01, 0.001]

ax3.scatter(stress_probs, stress_losses, s=100, alpha=0.7, label='Stress Tests', color='red')
ax3.scatter(mc_probs, mc_risk_levels, s=100, alpha=0.7, label='Monte Carlo', color='blue')

ax3.set_xlabel('Probability')
ax3.set_ylabel('Expected Loss')
ax3.set_title('Risk vs Probability: Stress Tests vs Monte Carlo')
ax3.set_xscale('log')
ax3.legend()
ax3.grid(True, alpha=0.3)

# 4. Risk Decomposition
# Show contribution of each asset to portfolio risk
asset_contributions = []
asset_names = list(portfolio.keys())

for asset in asset_names:
    weight = portfolio[asset]
    volatility = asset_params[asset]['volatility']
    contribution = weight * volatility  # Simplified risk contribution
    asset_contributions.append(contribution)

# Normalize to percentages
total_contribution = sum(asset_contributions)
asset_percentages = [c/total_contribution * 100 for c in asset_contributions]

colors = plt.cm.Set3(np.linspace(0, 1, len(asset_names)))
wedges, texts, autotexts = ax4.pie(asset_percentages, labels=asset_names, autopct='%1.1f%%', 
                                   colors=colors, startangle=90)
ax4.set_title('Risk Contribution by Asset')

plt.tight_layout()
plt.show()

print("📊 Stress testing and Monte Carlo analysis visualization complete")
```
